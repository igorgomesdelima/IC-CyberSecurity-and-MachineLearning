```python
# 0. Garante que o nfstream está instalado nesta sessão
!pip install nfstream

import urllib.request
from nfstream import NFStreamer
import pandas as pd

# 1. Usando um arquivo PCAP de teste super estável (oficial do projeto Wireshark)
url_pcap = "https://raw.githubusercontent.com/wireshark/wireshark/master/test/captures/dhcp.pcap"
urllib.request.urlretrieve(url_pcap, "pcap_teste_wireshark.pcap")
print("Download do PCAP concluído com sucesso!")

# 2. O NFStream lê o PCAP e transforma em Tabela (DataFrame)
meu_streamer = NFStreamer(source="pcap_teste_wireshark.pcap")
df_gerado_pelo_nfs = meu_streamer.to_pandas()

# 3. Mostramos as linhas geradas
df_gerado_pelo_nfs.head()

# 4. Exporta a tabela (DataFrame) que o NFStream gerou para um arquivo CSV real
df_gerado_pelo_nfs.to_csv("fluxos_wireshark_nfstream.csv", index=False)
print("Arquivo CSV gerado com sucesso!")
```
